```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Step 1: Load Dataset
file_path = 'city_day.csv'
df = pd.read_csv(file_path)

# Step 2: Data Overview
print(df.head())
print("\nMissing values per column:")
print(df.isnull().sum())

# Step 3: Handle Missing Values (simple forward fill)
df.fillna(method='ffill', inplace=True)

# Step 4: Convert date column if available
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])

# Step 5: Select pollutant features
dependent_vars = ['PM2.5', 'PM10', 'NO2', 'CO', 'O3']
feature_cols = [col for col in dependent_vars if col in df.columns]

# Step 6: Normalize features
scaler = MinMaxScaler()
df[feature_cols] = scaler.fit_transform(df[feature_cols])

# Step 7: Create AQI Category
def categorize_aqi(aqi):
    if aqi <= 50:
        return 'Good'
    elif aqi <= 100:
        return 'Moderate'
        #####dl
    elif aqi <= 200:
        return 'Poor'
    else:
        return 'Hazardous'

if 'AQI' in df.columns:
    df['AQI_Category'] = df['AQI'].apply(categorize_aqi)

# Final preview
print(df.head())

# Save file
df.to_csv('preprocessed_airquality_data.csv', index=False)
```
