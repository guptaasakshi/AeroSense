```python
# ======================================
# FP-GROWTH FREQUENT POLLUTION PATTERNS
# ======================================

import pandas as pd
from mlxtend.frequent_patterns import fpgrowth, association_rules

# --------------------------
# 1. Load Dataset
# --------------------------
df = pd.read_csv("city_day.csv")

pollutants = ["PM2.5", "PM10", "NO2", "SO2", "CO", "O3"]

# --------------------------
# 2. Convert to boolean format
# High pollution → 1 / Low → 0
# --------------------------
thresholds = {}

for col in pollutants:
    thresholds[col] = df[col].mean()     # mean cutoff
    df[col + "_HIGH"] = (df[col] > thresholds[col]).astype(int)

boolean_df = df[[c + "_HIGH" for c in pollutants]]

print("Boolean Data:")
print(boolean_df.head())

# --------------------------
# 3. FP-Growth Frequent Itemsets
# --------------------------
frequent_items = fpgrowth(boolean_df, min_support=0.1, use_colnames=True)

print("\nFrequent Pollution Patterns:")
print(frequent_items)

# --------------------------
# 4. Generate Association Rules
# --------------------------
rules = association_rules(frequent_items, metric="lift", min_threshold=1.0)

rules = rules.sort_values("lift", ascending=False)

print("\nAssociation Rules:")
print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])

# --------------------------
# 5. Save output
# --------------------------
frequent_items.to_csv("fp_growth_patterns.csv", index=False)
rules.to_csv("fp_growth_rules.csv", index=False)

print("\nSaved:")
print("fp_growth_patterns.csv")
print("fp_growth_rules.csv")
```
