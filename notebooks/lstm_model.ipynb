```python
###############################################
# LSTM 7-DAY AQI FORECAST - FULL NOTEBOOK CODE
###############################################

# =============================
# 1. IMPORTS
# =============================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
import joblib
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from sklearn.metrics import mean_absolute_error, mean_squared_error


# =============================
# 2. CONFIGURATION
# =============================
CSV_PATH = "city_day.csv"          # Your dataset
MODEL_OUT = "lstm_aqi_model.h5"
SCALER_OUT = "lstm_scaler.pkl"

SEQ_LEN = 30                       # How many past days model will look at
EPOCHS = 30
BATCH_SIZE = 16
TEST_SIZE = 0.2

USE_MULTIVARIATE = True            # True => AQI + pollutants, False => only AQI
DATE_COL = "Date"
TARGET_COL = "AQI"

POLLUTANTS = ["PM2.5", "PM10", "NO2", "CO", "O3", "SO2"]


# =============================
# 3. LOAD DATA
# =============================
df = pd.read_csv(CSV_PATH)
df[DATE_COL] = pd.to_datetime(df[DATE_COL])
df = df.sort_values(DATE_COL).reset_index(drop=True)

print("Raw dataset:", df.shape)
display(df.head())


# =============================
# 4. DAILY AGGREGATION
# =============================
num_cols = df.select_dtypes("number").columns.tolist()
df_daily = df.groupby(DATE_COL)[num_cols].mean().reset_index()

# Select features
if USE_MULTIVARIATE:
    features = [TARGET_COL] + [c for c in POLLUTANTS if c in df_daily.columns]
else:
    features = [TARGET_COL]

print("Using features:", features)

df_daily = df_daily[[DATE_COL] + features]


# =============================
# 5. HANDLE MISSING VALUES
# =============================
df_daily = df_daily.sort_values(DATE_COL).reset_index(drop=True)
df_daily[features] = df_daily[features].ffill().bfill()

print("Cleaned data:")
display(df_daily.head())


# =============================
# 6. SCALE DATA
# =============================
data = df_daily[features].values.astype(float)

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)
joblib.dump(scaler, SCALER_OUT)

print("Scaler saved:", SCALER_OUT)


# =============================
# 7. CREATE SEQUENCES
# =============================
def create_sequences(data_arr, seq_len):
    X, y = [], []
    for i in range(len(data_arr) - seq_len):
        X.append(data_arr[i:i + seq_len])
        y.append(data_arr[i + seq_len, 0])  # predicted target = AQI
    return np.array(X), np.array(y)

X, y = create_sequences(scaled_data, SEQ_LEN)

print("X shape:", X.shape)
print("y shape:", y.shape)


# =============================
# 8. TRAIN-TEST SPLIT
# =============================
split = int(len(X) * (1 - TEST_SIZE))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

print("Train samples:", X_train.shape[0])
print("Test samples:", X_test.shape[0])


# =============================
# 9. BUILD LSTM MODEL
# =============================
n_features = X_train.shape[2]

model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(SEQ_LEN, n_features)),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.1),
    Dense(16, activation="relu"),
    Dense(1, activation="linear")
])

model.compile(optimizer="adam", loss="mse", metrics=["mae"])
model.summary()


# =============================
# 10. TRAIN MODEL
# =============================
es = EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
mc = ModelCheckpoint("best_lstm.h5", save_best_only=True)

history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[es, mc],
    verbose=2
)

model.save(MODEL_OUT)
print("Model saved:", MODEL_OUT)


# =============================
# 11. TRAINING LOSS PLOT
# =============================
plt.figure(figsize=(8,4))
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.legend()
plt.title("LSTM Training Loss")
plt.show()


# =============================
# 12. TEST EVALUATION + INVERSE TRANSFORM
# =============================
def inverse_transform_aqi(scaled_vals, scaler_obj, n_features):
    dummy = np.zeros((len(scaled_vals), n_features))
    dummy[:,0] = scaled_vals.flatten()
    inv = scaler_obj.inverse_transform(dummy)
    return inv[:,0]

y_pred_scaled = model.predict(X_test)

y_test_real = inverse_transform_aqi(y_test.reshape(-1,1), scaler, n_features)
y_pred_real = inverse_transform_aqi(y_pred_scaled, scaler, n_features)

print("MAE:", mean_absolute_error(y_test_real, y_pred_real))
print("RMSE:", np.sqrt(mean_squared_error(y_test_real, y_pred_real)))

plt.figure(figsize=(10,4))
plt.plot(y_test_real, label="Actual AQI")
plt.plot(y_pred_real, label="Predicted AQI")
plt.legend()
plt.title("Test: Actual vs Predicted AQI")
plt.show()


# =============================
# 13. 7-DAY FUTURE FORECAST
# =============================
def forecast_next_days(model, scaler_obj, recent_data, days, seq_len, n_features):
    curr = recent_data.copy()
    preds_scaled = []

    for _ in range(days):
        prediction = model.predict(curr.reshape(1, seq_len, n_features))[0][0]
        preds_scaled.append(prediction)

        new_row = np.zeros(n_features)
        new_row[0] = prediction
        curr = np.vstack([curr[1:], new_row])

    preds_real = inverse_transform_aqi(np.array(preds_scaled).reshape(-1,1), scaler_obj, n_features)
    return preds_real


recent_scaled = scaled_data[-SEQ_LEN:]
pred_7 = forecast_next_days(model, scaler, recent_scaled, 7, SEQ_LEN, n_features)

print("7-day forecast:", np.round(pred_7, 2))


# =============================
# 14. FORECAST GRAPH
# =============================
past_dates = df_daily[DATE_COL].values[-SEQ_LEN:]
future_dates = pd.date_range(start=past_dates[-1] + pd.Timedelta(days=1), periods=7)

plt.figure(figsize=(10,5))
plt.plot(past_dates, df_daily[TARGET_COL].values[-SEQ_LEN:], label="Recent AQI")
plt.plot(future_dates, pred_7, marker='o', label="7-Day Forecast")
plt.xticks(rotation=45)
plt.legend()
plt.title("AQI Forecast for Next 7 Days")
plt.show()


# =============================
# 15. SAVE FORECAST CSV
# =============================
out = pd.DataFrame({"Date": future_dates, "Predicted_AQI": np.round(pred_7,2)})
out.to_csv("7day_aqi_forecast.csv", index=False)

print("Saved: 7day_aqi_forecast.csv")
display(out)
```
